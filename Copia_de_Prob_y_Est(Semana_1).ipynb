#Funciones principales:

1. **Estimación de parámetros: se basan en la información proporcionada por la muestra**



2. **pruebas de hipótesis: Se formulan hipótesis sobre la población y se utilizan datos muéstrales para evaluar la evidencia a favor o en contra de estas hipótesis.**



3. **Intervalos de confianza: Se construyen intervalos de valores que tienen una alta probabilidad de contener el verdadero valor del parámetro poblacional.**



4. **Modelado estadístico: Se utilizan modelos matemáticos para representar la relación entre variables en la poblacional.**

#Control de calidad del software:

* Ejemplo práctico:

***Durante las pruebas de un software de facturación, se registraron 25 defectos distribuidos en tres módulos:
Módulo de Cálculos: 15 defectos.
Módulo de Reportes: 7 defectos.
Módulo de Exportación: 3 defectos.
Conclusión:* Se priorizó la corrección del módulo de cálculos, lo que redujo defectos en un 60%***

***Conclusion:***  **De los 3 módulos, el módulo de cálculos representa más de la mitad de los módulos, y esto hace que los defectos bajen un 60%, teniendo que priorizar más a este primer módulo por representar más defectos que los demás**
